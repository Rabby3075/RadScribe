# Phase 1 EDA

This notebook checks the final Phase 1 manifest after label cleanup.

```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

PROJECT_ROOT = Path('..').resolve()
MANIFEST_PATH = PROJECT_ROOT / 'data' / 'processed' / 'manifest.parquet'
SPLIT_LABEL_COUNTS_PATH = PROJECT_ROOT / 'data' / 'processed' / 'split_label_counts.csv'

df = pd.read_parquet(MANIFEST_PATH)
df.shape
```

```python
df.head()
```

## Split Sizes

The split is grouped by `patient_id`, so images from the same study stay together.

```python
split_table = (
    df['split']
    .value_counts()
    .rename_axis('split')
    .reset_index(name='rows')
)
split_table['percent'] = (split_table['rows'] / len(df) * 100).round(1)
split_table
```

```python
split_table.plot(x='split', y='rows', kind='bar', legend=False, title='Split sizes')
plt.xlabel('split')
plt.ylabel('rows')
plt.show()
```

## Label Counts

Labels come from `Problems` first, then `MeSH`, then report text only as a fallback.

```python
label_table = (
    df.explode('labels')['labels']
    .value_counts()
    .rename_axis('label')
    .reset_index(name='rows')
)
label_table['percent'] = (label_table['rows'] / len(df) * 100).round(1)
label_table
```

```python
label_table.plot(x='label', y='rows', kind='bar', legend=False, title='Overall label counts')
plt.xlabel('label')
plt.ylabel('rows')
plt.show()
```

```python
split_label_counts = pd.read_csv(SPLIT_LABEL_COUNTS_PATH)
split_label_counts.pivot(index='labels', columns='split', values='count').fillna(0).astype(int)
```

```python
no_finding_pct = label_table.loc[label_table['label'] == 'No Finding', 'percent'].iloc[0]
other_pct = label_table.loc[label_table['label'] == 'Other', 'percent'].iloc[0]
rare = label_table[label_table['label'].isin(['Edema', 'Pneumothorax'])][['label', 'percent']]

print(f'No Finding: {no_finding_pct}%')
print(f'Other: {other_pct}%')
for _, row in rare.iterrows():
    print(f"{row['label']}: {row['percent']}%")
```

## Report Lengths

```python
report_lengths = df['report_text'].fillna('').str.split().str.len()
report_lengths.plot(kind='hist', bins=40, title='Report length in words')
plt.xlabel('words')
plt.show()
report_lengths.describe()
```

## Leakage Check

```python
max_splits_per_patient = df.groupby('patient_id')['split'].nunique().max()
assert max_splits_per_patient == 1
print('No patient appears in more than one split: no leakage.')
```

## Sample Image And Report Pairs

```python
sample = df.sample(3, random_state=7)

for _, row in sample.iterrows():
    relative_image_path = Path(str(row['image_path']).replace('\\', '/'))
    img = Image.open(PROJECT_ROOT / relative_image_path)
    plt.imshow(img, cmap='gray')
    plt.axis('off')
    plt.title(', '.join(row['labels']))
    plt.show()
    print(row['report_text'])
    print('-' * 80)
```

## Takeaway

This Phase 1 dataset has one cleaned frontal chest X-ray image per manifest row, linked to report text and train/val/test split.

Labels come from `Problems` first, then `MeSH`, then report text as fallback. The `Other` bucket is intentional: it holds real but out-of-scope findings such as calcified granulomas, spine changes, support hardware, and catheters.

The class balance is uneven, with large `Other` and `No Finding` groups and rare findings like Edema and Pneumothorax. Phase 2 should report per-class metrics, not only one overall score.
